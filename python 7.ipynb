### Question 1: What is a RESTful API?

**Answer:**
A RESTful API (Representational State Transfer API) is a web service that follows REST principles to allow communication between client and server using HTTP. It treats data as resources and uses standard HTTP methods like GET, POST, PUT, and DELETE to perform operations.

---

### Question 2: What is Flask, and why is it popular for building APIs?

**Answer:**
Flask is a lightweight Python web framework used to build web applications and APIs. It is popular because it is simple, flexible, easy to learn, and allows developers to build APIs quickly with minimal setup.

---

### Question 3: What are HTTP methods used in RESTful APIs?

**Answer:**
Common HTTP methods include:

* **GET** – Retrieve data
* **POST** – Create new data
* **PUT** – Update existing data
* **DELETE** – Remove data

---

### Question 4: What is the purpose of the @app.route() decorator in Flask?

**Answer:**
In Flask, `@app.route()` is used to define a URL endpoint and map it to a Python function. When a user accesses that URL, the associated function is executed.

---

### Question 5: What is the role of Flask-SQLAlchemy?

**Answer:**
Flask-SQLAlchemy is an extension that integrates SQLAlchemy with Flask. It simplifies database operations by providing an ORM (Object Relational Mapping) to interact with databases using Python objects instead of raw SQL.

---

### Question 6: How do you create a basic Flask application?

**Answer:**
A basic Flask app involves creating an instance of Flask and defining routes:

```python
from flask import Flask

app = Flask(__name__)

@app.route('/')
def home():
    return "Hello, World!"

if __name__ == '__main__':
    app.run(debug=True)
```

---

### Question 7: How do you return JSON responses in Flask?

**Answer:**
You can return JSON using `jsonify()`:

```python
from flask import jsonify

@app.route('/data')
def get_data():
    return jsonify({"name": "Alice", "age": 25})
```

---

### Question 8: How do you handle POST requests in Flask?

**Answer:**
Use the `methods` parameter and access request data:

```python
from flask import request

@app.route('/submit', methods=['POST'])
def submit():
    data = request.json
    return {"received": data}
```

---

### Question 9: How do you handle errors in Flask (e.g., 404)?

**Answer:**
You can define error handlers:

```python
@app.errorhandler(404)
def not_found(error):
    return "Page not found", 404
```

---

### Question 10: How do you structure a Flask app using Blueprints?

**Answer:**
Blueprints help organize large applications into modules:

```python
from flask import Blueprint

main = Blueprint('main', __name__)

@main.route('/')
def home():
    return "Home Page"
```

Then register it in the main app:

```python
app.register_blueprint(main)
```

This improves scalability and code organization.
